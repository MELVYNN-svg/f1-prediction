```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

event_code = "round_04"
driver = "VER"
pit_loss = 22.0

laps = catalog.load("fastf1_laps")
events = catalog.load("fastf1_events")

df = laps.loc[
    (laps["Driver"] == driver)
    & (laps["session_name"] == "race")
    & (laps["event_code"] == event_code)
    & (laps["Compound"].isin(["SOFT", "MEDIUM", "HARD"]))
    & (laps["LapTime"].notna())
    & (laps["PitInTime"].isna())
    & (laps["PitOutTime"].isna())
    & (laps["TrackStatus"] == "1")
    & (laps["LapNumber"] > 1)
].copy()

if "IsAccurate" in df.columns:
    df = df.loc[df["IsAccurate"].fillna(False)]

df["lap_seconds"] = df["LapTime"].dt.total_seconds()
df["TyreLife"] = pd.to_numeric(df["TyreLife"], errors="coerce")
df = df.loc[df["TyreLife"].notna()]

race_laps = int(df["LapNumber"].max())

profiles = (
    df.groupby(["Compound", "TyreLife"], as_index=False)["lap_seconds"]
    .mean()
    .sort_values(["Compound", "TyreLife"])
)

def make_profile(compound):
    sub = profiles.loc[profiles["Compound"] == compound].copy()
    x = sub["TyreLife"].to_numpy(dtype=float)
    y = sub["lap_seconds"].to_numpy(dtype=float)

    if len(x) < 2:
        return None

    x_unique, idx = np.unique(x, return_index=True)
    y_unique = y[idx]

    def f(tyre_life):
        tyre_life = np.asarray(tyre_life, dtype=float)
        return np.interp(tyre_life, x_unique, y_unique, left=y_unique[0], right=y_unique[-1])

    return f

compound_models = {
    "SOFT": make_profile("SOFT"),
    "MEDIUM": make_profile("MEDIUM"),
    "HARD": make_profile("HARD")
}

strategies = [
    ("SOFT", "MEDIUM"),
    ("SOFT", "HARD"),
    ("MEDIUM", "HARD"),
    ("MEDIUM", "SOFT"),
    ("HARD", "MEDIUM"),
    ("HARD", "SOFT")
]

def simulate_strategy(start_compound, next_compound, stop_lap, total_laps, pit_loss):
    f1 = compound_models[start_compound]
    f2 = compound_models[next_compound]

    if f1 is None or f2 is None:
        return np.nan

    first_stint_laps = np.arange(1, stop_lap + 1)
    second_stint_laps = np.arange(1, total_laps - stop_lap + 1)

    t1 = f1(first_stint_laps).sum()
    t2 = f2(second_stint_laps).sum()

    return t1 + pit_loss + t2

pit_laps = np.arange(1, race_laps)

plt.figure(figsize=(10, 6))

for c1, c2 in strategies:
    y = [simulate_strategy(c1, c2, lap, race_laps, pit_loss) for lap in pit_laps]
    plt.plot(pit_laps, y, marker="o", markersize=3, label=f"{c1} → {c2}")

event_name = (
    events.loc[events["event_code"] == event_code, "EventName"]
    .dropna()
    .iloc[0]
    if not events.loc[events["event_code"] == event_code, "EventName"].dropna().empty
    else event_code
)

plt.xlabel("Tour du pit stop")
plt.ylabel("Temps total de course estimé (s)")
plt.title(f"{driver} — {event_name} — stratégies 1 arrêt")
plt.legend()
plt.grid(True)
plt.show()
```

```python

import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time

YEAR = 2025
API = "https://api.openf1.org/v1"

http = requests.Session()

def fetch(endpoint, **params):
    r = http.get(
        f"{API}/{endpoint}",
        params=params,
        timeout=40
    )
    r.raise_for_status()
    return r.json()

sessions = pd.DataFrame(
    fetch("sessions", year=YEAR, session_name="Race")
)

sessions = (
    sessions.loc[
        (sessions["session_type"] == "Race")
        & (~sessions["is_cancelled"].fillna(False))
    ]
    .sort_values("date_start")
    .drop_duplicates("meeting_key")
    .reset_index(drop=True)
)

sessions["Manche"] = np.arange(1, len(sessions) + 1)

sessions["Grand Prix"] = (
    sessions["country_name"]
    + " - "
    + sessions["circuit_short_name"]
)

data = []

for _, s in sessions.iterrows():
    records = fetch("pit", session_key=int(s["session_key"]))

    if records:
        df = pd.DataFrame(records)
        df["Manche"] = s["Manche"]
        df["Grand Prix"] = s["Grand Prix"]
        data.append(df)

    time.sleep(2.1)

if not data:
    raise ValueError("Aucun arrêt trouvé.")

pits = pd.concat(data, ignore_index=True)

pits["stop_duration"] = pd.to_numeric(
    pits["stop_duration"], errors="coerce"
)

pits["lane_duration"] = pd.to_numeric(
    pits["lane_duration"], errors="coerce"
)

pits["driver_number"] = pd.to_numeric(
    pits["driver_number"], errors="coerce"
)

ver = pits.loc[pits["driver_number"] == 1].copy()

def calcul_moyennes(df, suffixe):
    return (
        df.groupby(["Manche", "Grand Prix"])
        .agg(
            **{
                f"Arrets_{suffixe}": ("lap_number", "size"),
                f"Stops_mesures_{suffixe}": (
                    "stop_duration", "count"
                ),
                f"Stop_moyen_{suffixe}": (
                    "stop_duration", "mean"
                ),
                f"Lane_moyen_{suffixe}": (
                    "lane_duration", "mean"
                )
            }
        )
        .reset_index()
    )

moy_ver = calcul_moyennes(ver, "VER")
moy_grille = calcul_moyennes(pits, "GRILLE")

resultat = (
    sessions[["Manche", "Grand Prix"]]
    .merge(moy_ver, how="left")
    .merge(moy_grille, how="left")
    .round(3)
)

tous_les_arrets = pits[
    [
        "Manche", "Grand Prix", "driver_number",
        "lap_number", "stop_duration", "lane_duration"
    ]
].sort_values(["Manche", "driver_number", "lap_number"])

arrets_verstappen = ver[
    [
        "Manche", "Grand Prix", "lap_number",
        "stop_duration", "lane_duration"
    ]
].sort_values(["Manche", "lap_number"])

display(resultat)
display(arrets_verstappen)
display(tous_les_arrets)

resultat.to_csv(
    "pitstop_moyennes_2025.csv", index=False
)

arrets_verstappen.to_csv(
    "pitstops_verstappen_2025.csv", index=False
)

tous_les_arrets.to_csv(
    "pitstops_tous_pilotes_2025.csv", index=False
)

for mesure, titre in [
    ("Stop_moyen", "Temps moyen d'immobilisation"),
    ("Lane_moyen", "Temps moyen dans la voie des stands")
]:
    fig, ax = plt.subplots(figsize=(16, 6))

    x = np.arange(len(resultat))
    w = 0.38

    ax.bar(
        x - w/2,
        resultat[f"{mesure}_VER"],
        w,
        label="Verstappen"
    )

    ax.bar(
        x + w/2,
        resultat[f"{mesure}_GRILLE"],
        w,
        label="Tous les pilotes"
    )

    ax.set_xticks(x)
    ax.set_xticklabels(
        resultat["Grand Prix"], rotation=75, ha="right"
    )

    ax.set_ylabel("Durée moyenne (secondes)")
    ax.set_title(f"{titre} - Saison {YEAR}")
    ax.legend()
    ax.grid(axis="y", alpha=0.3)

    plt.tight_layout()
    plt.show()
```
